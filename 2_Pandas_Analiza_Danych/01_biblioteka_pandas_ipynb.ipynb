# Series

```python
# tworzymy obiekt za pomoca funkcji Series , który automatycznie przyporzdkowywuje indeksy wierszy oraz kolumn
s = pd.Series([3, 2, 4 , 6 ] )
s
```

```python
# dodanie parametru ktory bedzie przyporzadkowywał wierszom nadane indexy
s = pd.Series([3, 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] )
s
```

```python
# dodanie parametru ktory bedzie przyporzadkowywał wierszom nadane indexy oraz nazwe naszej serii danych
s = pd.Series([3, 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] , name='nazwa_serii_danych' )
s
```

```python
# dostosowanie typu danych do podanego jak w numpy wszystkie musza miec ten sam typ danych
# jesli podamy rozne dane to seria przyjmie typ danych object
s = pd.Series([3., 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] )
s
```

! ! ! BRAKI DANYCH ! ! !  
jak sobie z nimi poradzić ?
w tym celu wykorzystujemy biblioteke numpy trzeba ja zaimportować


```python
import numpy as np
```

specjalny atrybut w bibliotece numpy który pozwala na przedstawianie brakujacych danych jako typ np.nan - określa brakujace dane

```python
np.nan
```

```python
# seria ze wstawionym typem przedstawiajacym brakujaca wartosc danych jako nan
s = pd.Series([3., np.nan, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] )
# brak wartosc w wierszu b
s
```

```python
# obiekt series składający sie z danych typu bool
s = pd.Series(data=[True , False , True ])
s
```

```python
# uzywanie danych i dzialanie z binlioteka dla danych czasowych z szeregami czasowymi itd czasem
# generowanie wartosci od 10 do 19 włacznie bez 20 | generujemy daty poczawszy od rok 2020-01-01 | parametr periods czyli ilosc dni od tego dnia
s = pd.Series(data=np.arange(10, 20) , index=pd.date_range(start='20200101' , periods=10))
s
```
