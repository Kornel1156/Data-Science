# Series

```python
# tworzymy obiekt za pomoca funkcji Series , który automatycznie przyporzdkowywuje indeksy wierszy oraz kolumn
s = pd.Series([3, 2, 4 , 6 ] )
s
```

```python
# dodanie parametru ktory bedzie przyporzadkowywał wierszom nadane indexy
s = pd.Series([3, 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] )
s
```

```python
# dodanie parametru ktory bedzie przyporzadkowywał wierszom nadane indexy oraz nazwe naszej serii danych
s = pd.Series([3, 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] , name='nazwa_serii_danych' )
s
```

```python
# dostosowanie typu danych do podanego jak w numpy wszystkie musza miec ten sam typ danych
# jesli podamy rozne dane to seria przyjmie typ danych object
s = pd.Series([3., 2, 4 , 6 ] , index = ['a', 'b', 'c', 'd' ] )
s
```

```python
# obiekt zawierajace infromacje o cenie notowana trzech społek ale jako slownik poprzez {} a nie liste []
price = pd.Series(data={'Apple' : 200 , 'CD Projekt' : 60 , 'Amazon' : 1900 })
price
```

```python
# wycinanie interesujacych danych ze danych
price['CD Projekt']
```

```python
# 2 sposob na wyciagniecie wartosci z tabeli danych obiektu poprzez obiekt.iloc[index] lub po prostu obiekt price[1]
print(price[1])
print(price.iloc[1])
```

```python
#zliczanie liczby elementow w naszym obiekcie
price.count()
```

```python
# funkcja value_counts() sprawdza ilosc wystapien poszczegolnych wartosci elementow w tabeli inaczej przedstawia rozklad , niezbedny np dla histogramu
price.value_counts()
```

Funkcje rowniez wystepujace w NumPy oraz w Pandas

```python
# funkcja sumujaca wartosci
price.sum()
```

```python
# max oraz min
print(price.min())
print(price.max())
```

```python
# odchylenie standardowe
price.std()
```

```python
# funkcja opisujaca obiekt zbiorczo przedstawiajaca statystyki
# ilosc , srednia , odchylenie . minimum , kwantyle rozkladu czyli 1/4 1/2 3/4 oraz max
price.describe()
```

```python
# dodajmy sobie index ktory nie ma wartosci
price = pd.Series(data={'Apple' : 200 , 'CD Projekt' : 60 , 'Amazon' : 1900 , 'KGHM'  : np.nan })
price
```

```python
# count nie uwzgledni tej danej
price.count()
```

```python
# wartoc tez nie zostanie uwzgledniona chyba ze ustawimy parametr dla funkcji dropna to wtedy zostana braki danych uwzglednione
price.value_counts(dropna=False)
```

```python
#funkcja do wyciagania indexow z najwiekszymi wartosciami *.nlargest(ilosc najwiekszych wartosci)
price.nlargest(2)
```

```python
#funkcja do wyciagania indexow z najmniejszymi wartosciami *.nsmallest(ilosc najmniejszymi wartosci)
price.nsmallest(2)
```

```python
# funkcja uporzadkowywujaca dane w tabeli
price.rank()
```

```python
# funkcja sortujaca indexy
price.sort_index()
```

```python
# funkcja sortujaca po wartosciach
price.sort_values()
```

```python
# chcac zmienic kolejnosc sortowania z domyslnego rosnacego to ustawiamy parametr ascending=False
price.sort_values(ascending=False)
```

```python
# metoda apply ktora przyjmue jako pierwszy parametr funkcje  , wybieramy lambda i mnozemy kazdy z elementow przez 3.8 na złotowki
price_pln  = price.apply(lambda x : x * 3.8)
price_pln
```
