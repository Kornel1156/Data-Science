### <a name='a1'></a> wczytanie Danych


```python
# pobranie danych z api finance yahoo
import yfinance as yf

def fetch_financial_data(company='AMZN'):
    """Pobiera dane dotyczące notowań spółki."""
    return yf.download(company)

apple = fetch_financial_data('AAPL')
amazon = fetch_financial_data('AMZN')
google = fetch_financial_data('GOOGL')
uber = fetch_financial_data('UBER')

# spłaszczenie kolumn w yfinance
apple.columns = apple.columns.get_level_values(0)
amazon.columns = amazon.columns.get_level_values(0)
google.columns = google.columns.get_level_values(0)
uber.columns = uber.columns.get_level_values(0)

```

```python
# wyswietlenie danych o notowaniach spółki

# Kazda tabela danych spółek ma inna ilosc wierszy ale łaczy ich index

uber.info()
```

```python
# 5 wierszy apple
apple.head()
```

### <a name='a2'></a> Preprocessing

```python
# przed polaczeniem danych nalezy przetworzyc nazwy kolumn
# dodanie przedrostka apple przez kazdym rekordem
apple.columns = ['apple_' + col.lower() for col in apple.columns ]
google.columns = ['google_' + col.lower() for col in google.columns ]
amazon.columns = ['amazon_' + col.lower() for col in amazon.columns ]
uber.columns = ['uber_' + col.lower() for col in uber.columns ]
```

```python
apple.head()
```

### <a name='a3'></a> konkatenacja danych

```python
# łaczenie danych  , tabele z danymi ktore chcemy połaczyc oraz wzgledem jakiej osi chcemy je łaczyć wybiermay os y
df = pd.concat(objs=[apple,amazon,google,uber] , axis =1)
```

```python
# wyłonienie statystyk
df.describe().T
```

```python
# formatownie danych dla lepszej reprezentacji
pd.set_option('display.float_format' , lambda x: f'{x:.2f}')
df.describe().T
```

```python
# wspołczynnik korelacji poszczegolnych zmiennych
# wspołczynnik korelacji - Współczynnik korelacji mówi, jak silnie dwie rzeczy poruszają się względem siebie.
df.corr()
```

```python
# przedstawienie wszystkich dostepnych kolumn
df.columns
```

```python
# policznie wspolcznika korealcji tylko dla wybranych np kolumn albo wycinka danych

# ceny zamkniecia - czyli rekordy dla danej kolumn ktorej nazwa konczy sie na close
closes = [col for col in df.columns if col.endswith('close')]
closes
```

```python
# wycinamy obiekt dataFrame ktora reprezentuje kolumny z cena zamkniecia

# ! ! !  UWAGA ! ! !
# dataframe - to obiekt ktory jest tabela ma wiersze i kolumny
# series - to obiekt ktory ma jedna kolumne'

print(df[closes])
print(type(df[closes]))
```

```python
# liczymy korelacje zmiennych
df[closes].corr()
```

```python
# wolumen obrotu
closes = [col for col in df.columns if col.endswith('close') or col.endswith('volume') ]
closes
```

```python
# wspołczynnik korelacji poszczegolnych zmiennych
# wspołczynnik korelacji - Współczynnik korelacji mówi, jak silnie dwie rzeczy poruszają się względem siebie.
df[closes].corr()
```
