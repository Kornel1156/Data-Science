### <a name='a6'></a> Obliczanie nowych kolumn

````
df = fetch_financial_data('UBER')
df.info()
````

````
df.head()
````

````
df.describe()
````

````
df.head(3)
````

````
df['Average'] = (df.Open + df.Close) / 2.0
df.head()
````

````
df = df.sort_index()
df.head()
````

````
df['Close_shift'] = df.Close.shift(1)
df.head()
````

````
df['Daily_Change'] = df.Close / df.Close_shift - 1
df.head()
````

````
del df['Daily Change']
````

````
df.head()
````

````
df.Daily_Change.min()
````

````
df.Daily_Change.max()
````
