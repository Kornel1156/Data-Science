### <a name='a6'></a> Obliczanie nowych kolumn

````
df = fetch_financial_data(company='UBER')
# spłaszczenie kolumny w yfinance
df.columns = df.columns.get_level_values(0)
df.info()
````

````
# 5 pierwszych kolumn
df.head()
````

````
# podstawowe statystyki
df.describe()
````

````
# pierwsze 3 rekordy
df.head(3)
````

````
# Utworzenie nowej kolumny czyli srednia artytmeryczna 2 kolumn
df['Average'] = (df.Open + df.Close) / 2.0
df.head()
````

````
# sortowanie indexow czyli wierszy ramki danych , przypisanie do zmiennej
df = df.sort_index()
df.head()
````

````
# przesuniecie wartosci kolumny o jeden w dol na pierwszym , miejscu wstawiony zostanie brak danych
# PRZYKŁADOWO DLA LICZENIA # DNIOWEJ STOP WZROTU
print(df[['Close']].shift(1))
print(df[['Close']].shift(3)) # jesli chcemy liczbyc stope zwrotu czyli jak sie zmieniala wartosc aktywa z 3 dni a nie codziennie jak sie zmieniala
````

````
# dodanie nowej kolumny
# TERAZ WYGODNIEJ JEST LICZYC DZIENNA ZMIANNE (STOPE PROCENTOWA) JESLI wartosc zamkniecia rynku z nastepnego dnia jest na tej samej wysokosci do
# wartosc z zamkniecia z dania wczesniejszego
df['Close_shift'] = df.Close.shift(1)
df.head()

# 2026-07-16	74.040001 / 72.669998 = ile wczorajsza cena w procentach(%) stanowi dzisiejsza cene np wczorajsza cena stanowi 90% ceny dzisiejszej

````

````
# dostajemy stope zwrotu w nowej kolumnie Daily_Change mówiąca ile stracilismy procent albo zarobilism w stosunku do kwoty
# ktora zainwestowalismy początkowo
df['Daily_Change'] = df.Close / df.Close_shift - 1
df.head()
````

### <a name='a7'></a> Filtrowanie danych

````
df
````

````
# sprwadzamy w które dni stopa zwrotu była wieksza od zera
# dodajemy kolumne z informacja prawdy czy tez falszu kiedy stopa zwrotu byla dodatnia a kiedy ujemna
df['df_positive'] =  df['Daily_Change'] > 0
````

````
df
````

````
# przypisanie do zmiennej
df_positive =  df[df.Daily_Change > 0 ]
df_positive
````

````
# policzenie sredniej dodatniej stopy zwrotu
df_positive.Daily_Change.mean()
````

````
 # ujemna ramka danych z rekordami ktorych stopa zwrotu była ujemna poprzez maske df[df.Daily_Change < 0]
df_negative = df[df.Daily_Change < 0]
df_negative
````

````
# średnia ujemna stopa zwrotu
df_negative.Daily_Change.mean()
````

````
# sprawdzamy czy w ciagu sesji zdazyło sie aby cena zamkniecia była równa najwyższej cenie
df.Close == df.High
````

````
# wyciagamy te rekordy ktore spelniaja nasza maske logiczna
df[df.Close == df.High]
````

````
# rekordy ktore byly rowne cenie najnizeszeje na cenie zamkniecia
df[df.Close == df.Low]
````

````
# sprawdzamy ktore  daty , indexy sa powyzej penwego moment
df.index > '2022-11-01'
````

````
# teraz wyświetlamy rekordy ktore spelniaja wlasnie ten wartunek czyli sa z odpowiedniego dnia
df[df.index > '2019-11-01']
````

````
# wyciagniecie tych rekordow ktore spelniaja okreslaone warunki (maske logiczna) czyli
df[(df.index >= '2026-07-15') & (df.index < '2026-07-27')]
````
