### <a name='a1'></a> import bibliotek




```python
# import bibliotek
import pandas as pd
print(pd.__version__)
```

### <a name='a2'></a> załadaowanie danych


```python
url = 'https://storage.googleapis.com/esmartdata-courses-files/ds-bootcamp/appstore_games.csv'
df_raw = pd.read_csv(url)
df_raw.head()
```

### <a name='a3'></a> preprocessing - przygotowanie danych


```python
# sprawdzenie jakie mamy kolumny
df_raw.columns
```

```python
# kopia obiektu aby nie brudzić obiektu głownego z danymi
df = df_raw.copy()
df.info()
```

```python
# usuniecie kolumn ktore nas nie interesuja poprzez parametr drop
df = df.drop(columns=['URL' , 'Icon URL' , 'Subtitle' , 'Description' , 'In-app Purchases'])
# po nieuwzglednieniu kilku kolumn
df.head()
```

```python
# ustawiamy indeksy rekordow jako osobne kolumny
df = df.set_index('ID')
df.head()
```

```python
# statystyki danych których ilosc kolumn jest mniejsza niz rzeczywista ilosc poniewaz zostały policzone tylko dla kolumn z danymi o typie flaot albo int alob double
df.describe()
```

```python
# zmiana kolumn z wierszami dla czytelnosi
df.describe().T
```

```python
# statystyki wybranych danych czyli zamiast statystyk dla kolumn z danymi o typie float oraz int jeszcze mamy dla danych z typem object
df.describe(include=['object'])
```

```python
# zmiana werszy z kolumną
df.describe(include=['object']).T
```

```python
# ustandaryzowanie danych które wystepuja czyli:
# 1. spacje znajdujace sie w nazwie zastąpić podkreślnikiem
# 2. ustandaryzujemy wszytkie literny jakiekolwiek duze z slowie damy na male
df.columns
```

```python
# ustandaryzowanie nazw kolumn
df.columns = [col.lower().replace(' ' , '_') for col in df.columns]
df.head()
```

```python
# dzieki podłagom mamy szybkie wyszukiwanie nazwy takiej kolumny poprzez podpowiedz
df.average_user_rating
```
