### Funkcje statystyczne

```
import numpy as np
np.__version__
```

```
# cena na giełdzie z ostanich 3 dni
# | open | hight | low | close | volumen |
price = np.array([[12.40, 12.80, 11.90, 12.60, 1000],
                  [12.50, 13.00, 11.70, 12.20, 2000],
                  [12.20, 13.40, 12.20, 13.20, 1500]])
print(price)
```

sum()

```
# suma wszystkich elementow ktore znajduja sie w tablicy
price.sum()
```

```
price.sum(axis=0)# sumuje w osi pionowej w osi Y czyli podaje sumy kolumn z osobna
```

```
price.sum(axis=1) # sumuje wartosci z wierszy X
```

```
np.sum(price) # całkowita suma naszej tabeli
```

min()

```
np.min(price, axis=0) # minimalna wartosc z kolumn z wymiaru kolumn y
```

max()

```
np.max(price, axis=0) # maksmyalne wartiscu ze wszystkuch kolumn
```

median()

```
np.median(price)
```

mean() # średnia wymiarow

```
np.mean(price, axis=0)
```

std() # miara rozrzutu czyli odchylenie standardowe . Odchylenie standardowe jest w tej samej skali i jednostce co dane. Jeśli mierzysz wzrost w centymetrach, odchylenie standardowe również jest wyrażone w centymetrach. Dzięki temu można powiedzieć np.:

„Wyniki odbiegają od średniej przeciętnie o około 5 cm.”

```
np.std(price, axis=0)
```

var() # Wariancja to miara rozproszenia danych. Mówi nam, jak bardzo wartości w zbiorze różnią się od średniej.
Wariancja nie jest w tej samej skali, ponieważ podczas jej obliczania odchylenia od średniej są podnoszone do kwadratu. Dlatego jednostką wariancji są cm², kg², punkty² itd. To sprawia, że jej interpretacja jest mniej intuicyjna.

```
np.var(price, axis=0)
```
