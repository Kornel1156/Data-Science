# Indeksowanie , wycinanie

```python
# generowanie danych
A = np.arange(20)
print(A)
```

```python
# jak wyciac konkretny element poprzez jego indeks zaczyna sie index od zera 0
# A[idx] , A[start : stop] , A[:stop] , A[start:]
```

```python
# chcac wyciąc 2 element to :
print(A[2])
print(A[2:])
print(A[:2]) # bez uwzglednienia elemntu stopu
```

```python
# wycinanie konkretnych elementow
A[[0,2]]
```

```python
# indeksowanie od konca
A[-1] # ostatnia wartosc
```

```python
# od 10(z zawartym 10 elementem)  do 15 elementu (nie zawierajac 15 elementu)
A[10:15]
```

```python
# przeksztalcanie tablicy
A = A.reshape(4,5)
A
```

```python
# wycianie konkretnych macierzy , kolumn czy wierszy z calosci
#cala kolumna
A[:,0] # [wszystkie_wiersze , 0 element ]
```

```python
# wyciecie fragmentu tablicy
A[1, 1:4]
```

```python
# modyfikacja wartosci
A[1:3 , 1:4] = 67
A
```

# Iteracja po tablicach

```python
# przechodzenie po wierszach poprzez petle
for row in A:
    print(row)
```

```python
# przechodzenie po wszystkich elementach , pierwsze elemty
for row in A:
  print(row[0])
```

```python
# trzy pierwsze elementy
for row in A:
  print(row[:3])
```

```python
# Iterownie po kazdym elemencie po spłaszczeniu
# flat zwraca iterator, który przechodzi przez wszystkie elementy tablicy
for item in A.flat:
  print(item)
```

# Maski Logiczne

```python
A = np.arange(start=-10 , stop=10 , step=0.5)
A = A.reshape(10,-1)
A
```

```python
# sprawdzanie w ktorych miejscach mamy wartosci ujemne
A > 0 # to rownianie zwraca maske logiczne prawda gdu > 0 falsz gdy < 0
```

```python
# wyciecie tych elementow ktore spelniaja
A[A > 0 ]
```

```python
# funkcja bitwise_and ktora pozwala na konstruowanie szerokich warunkow
np.bitwise_and(A > -5 , A < 5)
```

```python
# bitwise_or - mozmy podac dwa warunki
np.bitwise_or(A < -5 , A > 5)
```
