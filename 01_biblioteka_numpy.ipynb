# Tworzenie_Tablic

```python
np.zeros(shape=(4,10)) # tworzy dowolna tablica utworzoną z samych zer
```

Dodanie z okresleniem typow naszych wartosci w takiej tablicy

```python
np.zeros(shape=(4,10) , dtype='int') # tworzy dowolna tablica utworzoną z samych zer z typem
```

tablica z samymi jedynkami

```python
np.ones(shape=(5,5) ,  dtype='complex')
```

tworzenie tablic z wartosciami ktore my chcemy

```python
np.full(shape=(4,10) , fill_value=4 , dtype='double') # tworzy dowolna tablica utworzoną z samych zer z typem oraz nasza wartscia
```

wygeneruje dane w zakresie

```python
np.arange(10)
```

ustawienie poczatek koniec i co ile

```python
np.arange(start=5 , stop=12)
```

```python
np.arange(start=1 , stop=12 , step=3) # dodatkowy parametr definiujacy krok z jakim tworzona jest tablica
```
