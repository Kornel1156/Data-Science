# Podstawowe operacje na tablicach

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
print(a)
print(b)
```

```python
a + b # dodawanie element po elemencie
```

```python
a - b # odejmowanie element po elemencie
```

```python
a * b
```

```python
a / b  # uwaaga nie mozna dzielic przez 0 wiec tablica nie moze miec zera w mianowniku
```

```python
# dodanie 3 do kazdego elementu
a + 3
```

```python
# inne i łączone
a + 3*  b
```

```python
# rowniez metody ktore dodaja dwie tablice
np.add(a,b)
```

```python
# odejmowanie przez metode
np.subtract(a,b)
```

```python
# metoda mnożaca dwie tablice
np.multiply(a,b)
```

```python
# metoda do dzielenia
np.divide(a,b)
```

```python
# macierze
x = np.array([[1,3] , [-2 ,0]])
y = np.array([[6,0] , [-1 ,-2]])
print(x , '\n')
print(y)
# 1 * 6 + 3 * 0 itd ...
```

```python
# mnożenie macierzowe wiersz * kolumn
np.dot(x,y)
```

```python
# 2 spsopob na mnozenie macierzy
x.dot(y)
```

```python
# uwaga mnożenie macierzy nie jest przemienne
y.dot(y)
```

```python
# mnożenie macierzy z algebry liniowej na krzyz albo w zalensoci od rozmirow m x n
x @ y
```

# Podstawowe funkcje

```python
# all sprawdza kazdy elemnt i sprawdza czy kazda zwróci true to zwróci true wpp zwróci false
print(np.all([1,2,3,4]))
print(np.all([1,2,3,0]))
```

```python
# all sprawdza czy jakikolwiek elemnt jest true to zwróci true wpp zwróci false
print(np.any([1,2,3,4]))
print(np.all([1,2,3,0]))
```

```python
# funkcja bool dla 0 zwroci false a dla kazdej innej liczby zwróci true
bool(0)
```

```python
# funkcja generujaca wartosci z rozkładu normalnego 5 roznych wartosci
print(np.random.randn(5))
# funkcja generujaca wartosci z rozkładu jednostajnego 5 roznych wartosci
print(np.random.rand(5))
```

```python
# przyjmijmy ze wynik tego to prawdopodobienstwo otrzymania poszczegolnej klasy dla danej probki
# czyli mamy probelm klasyfikacji mamy 5 klas i chcemy je przyporzadkowac , model zwrocil wartosci prawdopodobienstw
A = np.random.rand(5)
print(A)
# sprobujmy uzyc funkcji ktora zwróci nam indeks elementu gdzie nasza tablica przyjmuje wartosi najwieksza
print(np.argmax(A))
# chcac wyciac tą wartosc z naszej tablicy użyjemy :
print(A[np.argmax(A)])
# równiez jest funkcja zwracająca wartosc najniższą
print(np.argmin(A))
# jezeli mamy klasyfikacje wieloklasową to mozemy posortowac prawdopodobienstawa ktora sortuje wskazujac indeksami elemntu od min do max poprzez :
# czyli jesli chcemy przedstawic wedle najwiekszego prawdopodobienstwa elemntu do danej klasyfikacji
print(np.argsort(A))
```

```python
# bardziej statystyczne
# maksimum
print(np.max(A))
# minimum
print(np.min(A))
```

```python
# srednia
np.mean(A)
```

```python
# mediana
np.median(A)
```

```python
# odchylenie standardowe
np.std(A)
```
